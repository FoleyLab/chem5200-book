## Computing the bra-ket
The bra-ket $\langle\phi|\psi\rangle$ (also called the inner product) is the **probability amplitude** for projecting $|\psi\rangle$ onto $|\phi\rangle$. The amplitude may be real or complex. For normalized states, its magnitude lies between 0 and 1, and the corresponding measurement probability is $|\langle\phi|\psi\rangle|^2$. Orthogonal states have zero overlap, while states that differ by at most an overall phase have overlap magnitude 1. The $S_z$ basis states are orthonormal, which is codified by

$$\langle \chi_n | \chi_m\rangle = \delta_{nm} $$

where we have used the Kronecker delta function $\delta_{nm} = 0$ if $n\neq m$ and $\delta_{nm} = 1$ if $n=m$ and
we are using $\chi_n$ and $\chi_m$ to represent arbitrary spin states.

With their vector representations, we can compute the bra-ket using the dot product as follows:

## 📈 Expectation Values

In quantum mechanics, an **expectation value** is the average result you'd expect to obtain if you repeatedly measured a quantum observable on a system in a given state. It is written as a **bra-ket sandwich**:

$$
\langle n | \hat{O} | m \rangle
$$

If the bra and ket are the **same** state (i.e. $|n\rangle = |m\rangle$), this gives the expectation value of operator $\hat{O}$ in that state. In NumPy, this looks like:

```python
expectation_value = bra @ Operator @ ket
```

## 🎯 Eigenvalues and Eigenvectors

An important property of **Hermitian matrices** is that all their **eigenvalues are real**.

In quantum mechanics, the eigenvalues of Hermitian operators represent the **possible outcomes of measurements**. In this notebook, we’ve been exploring the **spin angular momentum observable**, especially its **z-component** represented by the operator $ \hat{S}_z $.

We’ve already seen that $\hat{S}_z $ has two eigenstates:

- $ |\alpha\rangle $ with eigenvalue $ +\frac{\hbar}{2} $
- $ |\beta\rangle $ with eigenvalue $ -\frac{\hbar}{2} $

These relationships are preserved in our **matrix-vector** representation.

---

### 📐 What Are Eigenvectors?

For a matrix $ \mathbb{M} $, a **vector** $ \mathbf{x}$ is an **eigenvector** if:

$$
\mathbb{M} \mathbf{x} = \lambda \mathbf{x}
$$

Where:
- $ \lambda $ is the **eigenvalue** (a number),
- $\mathbf{x} $ is a **non-zero vector** that only gets **scaled** by the matrix—not rotated.

This is rare and special! Normally, matrices change both the **magnitude and direction** of vectors.

### 🔁 Commutators and Compatibility

Later in quantum mechanics, we’ll explore the **generalized uncertainty principle**. One of the most important mathematical tools in that formulation is the **commutator**.

---

### 🧮 What Is a Commutator?

For two operators $ \hat{A} $ and $ \hat{B} $, the **commutator** is defined as:

$$
[\hat{A}, \hat{B}] = \hat{A} \hat{B} - \hat{B} \hat{A}
$$

This same formula applies directly to **matrices** representing those operators.

---

### 🧠 Key Facts About Commutators

1. If $[\hat{A}, \hat{B}] = 0 $, we say that **$ \hat{A} $ and $ \hat{B} $ commute**.
2. If $ [\hat{A}, \hat{B}] \ne 0 $, the operators **do not commute**.
3. Two commuting Hermitian operators can be **simultaneously diagonalized**, so they possess a common orthonormal eigenbasis. This wording remains correct when either operator has degenerate eigenvalues.
4. **Commuting observables** are called **compatible observables**: they can have simultaneous definite values in a common eigenstate.
5. **Non-commuting operators** correspond to **incompatible observables**: they cannot, in general, have simultaneous definite values, and an uncertainty relation constrains their joint precision.

---

### 📐 Commutation Relations of Spin Operators

The spin operators obey the following famous commutation relations:

$$
[\hat{S}_x, \hat{S}_y] = i\hbar \hat{S}_z
$$
$$
[\hat{S}_y, \hat{S}_z] = i\hbar \hat{S}_x
$$
$$
[\hat{S}_z, \hat{S}_x] = i\hbar \hat{S}_y
$$

These relations tell us that **no pair of spin components is compatible** — there's inherent uncertainty in knowing any two simultaneously.

## From the Hückel Model to an H$_2^+$ Potential Energy Surface

The simple matrix above assumes an orthonormal basis. Real hydrogen 1s orbitals on different nuclei overlap, so

$$S(R)=\langle 1s_A|1s_B\rangle \ne 0.$$

We must therefore solve the generalized eigenvalue problem

$$\mathbf{H}(R)\mathbf{c}=E_{\mathrm{elec}}(R)\,\mathbf{S}(R)\mathbf{c},$$

where

$$
\mathbf{H}(R)=\begin{bmatrix}\alpha_{\mathrm{H}}(R)&\beta_{\mathrm{H}}(R)\\\beta_{\mathrm{H}}(R)&\alpha_{\mathrm{H}}(R)\end{bmatrix}, \qquad
\mathbf{S}(R)=\begin{bmatrix}1&S(R)\\S(R)&1\end{bmatrix}.
$$

The symmetry-adapted bonding and antibonding electronic energies are

$$E_{\mathrm{bond}}^{\mathrm{elec}}(R)=\frac{\alpha_{\mathrm{H}}(R)+\beta_{\mathrm{H}}(R)}{1+S(R)}, \qquad
E_{\mathrm{antibond}}^{\mathrm{elec}}(R)=\frac{\alpha_{\mathrm{H}}(R)-\beta_{\mathrm{H}}(R)}{1-S(R)}.$$

A molecular potential energy surface must also include the proton-proton repulsion, $1/R$, so $V(R)=E_{\mathrm{elec}}(R)+1/R$. All quantities below use atomic units: $R$ is in bohr and energy is in hartree.


````python
def overlap_S(R):
    """Return the overlap integral between the two hydrogen 1s orbitals."""
    return (1.0 + R + (R**2) / 3.0) * np.exp(-R)


def coulomb_alpha(R):
    """Return the H2+ Coulomb matrix element alpha_H(R)."""
    J = (1.0 / R) - ((1.0 / R) + 1.0) * np.exp(-2.0 * R)
    return -0.5 - J


def resonance_beta(R):
    """Return the H2+ resonance matrix element beta_H(R)."""
    K = (1.0 + R) * np.exp(-R)
    return -0.5 * overlap_S(R) - K


def h2plus_potential_energies(R):
    """Return the bonding and antibonding H2+ energies, including 1/R."""
    S = overlap_S(R)
    alpha_H_R = coulomb_alpha(R)
    beta_H_R = resonance_beta(R)

    E_bond_electronic = (alpha_H_R + beta_H_R) / (1.0 + S)
    E_antibond_electronic = (alpha_H_R - beta_H_R) / (1.0 - S)

    nuclear_repulsion = 1.0 / R
    V_bond = E_bond_electronic + nuclear_repulsion
    V_antibond = E_antibond_electronic + nuclear_repulsion
    return V_bond, V_antibond
````

### 📈 Demonstration: bonding and antibonding potential energy curves

The following cell evaluates both states over a range of internuclear distances. The dotted line at $-0.5$ hartree is the H $+$ H$^+$ dissociation limit.


````python
import matplotlib.pyplot as plt

R_values = np.linspace(0.5, 8.0, 500)
V_bond, V_antibond = h2plus_potential_energies(R_values)

minimum_index = np.argmin(V_bond)
R_equilibrium = R_values[minimum_index]
V_minimum = V_bond[minimum_index]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(R_values, V_bond, lw=2, color="royalblue", label=r"bonding $\sigma_g$")
ax.plot(R_values, V_antibond, lw=2, color="crimson", label=r"antibonding $\sigma_u^*$")
ax.axhline(-0.5, color="black", ls=":", label=r"H + H$^+$ dissociation limit")
ax.plot(R_equilibrium, V_minimum, "o", color="royalblue")
ax.annotate(
    rf"minimum: $R \approx {R_equilibrium:.2f}$ bohr",
    xy=(R_equilibrium, V_minimum),
    xytext=(R_equilibrium + 0.6, V_minimum - 0.10),
    arrowprops={"arrowstyle": "->", "color": "royalblue"},
)
ax.set_xlabel(r"internuclear distance $R$ (bohr)")
ax.set_ylabel(r"total energy $V(R)$ (hartree)")
ax.set_title(r"LCAO potential energy curves for H$_2^+$")
ax.set_ylim(-0.7, 2.0)
ax.legend()
ax.grid(alpha=0.2)
plt.show()

print(f"Approximate bonding minimum: R = {R_equilibrium:.3f} bohr, V = {V_minimum:.6f} hartree")
````
