```python
import py3Dmol
from pyscf import gto, scf
from pyscf.geomopt.geometric_solver import optimize

# 1. Define the initial unoptimized molecular geometry (in Angstroms)
# Here we use an intentionally distorted water molecule to watch it relax
mol = gto.M(
    atom='''
        O   0.000000   0.000000   0.000000
        H   0.000000   1.200000   0.200000
        H   0.900000  -0.500000   0.500000
    ''',
    basis='6-31g',
    unit='Angstrom'
)

# 2. Set up the Mean-Field (Hartree-Fock) calculation
mf = scf.RHF(mol)

# 3. Specify an output file path where geomeTRIC will record the trajectory steps
trajectory_xyz = 'optimization_trajectory.xyz'

# 4. Run the geometry optimization
# We pass 'xyzout' to save the coordinates of every step into an XYZ file
mol_eq = optimize(mf, maxsteps=100, xyzout=trajectory_xyz)

print("\n--- Optimization Complete ---")
print("Final Optimized Coordinates (Bohr):")
print(mol_eq.atom_coords())

# 5. Read the recorded trajectory and render the interactive animation
with open(trajectory_xyz, 'r') as f:
    trajectory_data = f.read()

# Initialize the py3Dmol viewer with the full multi-frame XYZ trajectory
viewer = py3Dmol.view(width=400, height=400)
viewer.addModelsAsFrames(trajectory_data, 'xyz')

# Apply styles: Stick representation for bonds, spheres for atoms
viewer.setStyle({'stick': {}, 'sphere': {'scale': 0.3}})

# Zoom to fit the molecule layout
viewer.zoomTo()

# Configure the visualizer to animate through the optimization steps
# 'loop': 'backward' smoothly rewinds, 'interval': 200 is speed in milliseconds
viewer.animate({'loop': 'backward', 'interval': 200})

# Render the widget inside the Colab notebook output
viewer.show()
```
